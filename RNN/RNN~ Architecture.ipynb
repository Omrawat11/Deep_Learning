```python
import tensorflow as tf
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense, SimpleRNN
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
model = Sequential()

model.add(SimpleRNN(3,input_shape=(4,5)))
model.add(Dense(1,activation='sigmoid'))

model.summary()
```

```python
print(model.get_weights()[0].shape)
model.get_weights()[0]
```

```python
# Get the weight matrix
weight_matrix_0 = model.get_weights()[0]

# Create a heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(weight_matrix_0, annot=True, cmap='viridis', fmt=".2f")
plt.title('Weight Matrix Visualization (Input to SimpleRNN)')
plt.xlabel('Hidden Units')
plt.ylabel('Input Features')
plt.show()
```

```python
print(model.get_weights()[1].shape)
model.get_weights()[1]
```

```python
weight_matrix_1 = model.get_weights()[1]

plt.figure(figsize=(8,6))
sns.heatmap(weight_matrix_1, annot=True, cmap='viridis', fmt=".2f")
plt.title('Weight Matrix Visualization (SimpleRNN to Output)')
plt.xlabel('Hidden Units')
plt.ylabel('Input Features')
plt.show()
```
