```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense,LSTM,GRU
```

```python
# Load the IMDb dataset
(x_train, y_train), (x_test, y_test) = imdb.load_data(num_words=10000)

# Pad sequences to have the same length
x_train = pad_sequences(x_train, maxlen=100)
x_test = pad_sequences(x_test, maxlen=100)
```

```python
from tensorflow.keras.layers import Input
from tensorflow.keras.utils import plot_model
from IPython.display import Image, display

# Define the SimpleRNN model architecture using Input layer
rnn_model = Sequential([
    Input(shape=(100,)),
    Embedding(10000, 32),
    SimpleRNN(5, return_sequences=True),
    SimpleRNN(5),
    Dense(1, activation='sigmoid')
])

# Generate and display the visualization
plot_model(rnn_model, show_shapes=True, show_layer_names=True, to_file='rnn_model.png')
display(Image(filename='rnn_model.png'))
```

```python
print(rnn_model.summary())
```

```python
## FOR LSTM model (Deep LSTM)
LSTM_model = Sequential([
    Input(shape=(100,)),
    Embedding(10000, 32),
    LSTM(5, return_sequences=True),
    LSTM(5),
    Dense(1, activation='sigmoid')
])

# Generate and display the visualization
plot_model(LSTM_model, show_shapes=True, show_layer_names=True, to_file='LSTM_model.png')
display(Image(filename='LSTM_model.png'))
```

```python
print(LSTM_model.summary())
```

```python
## FOR GRU model (Deep GRU)
GRU_model = Sequential([
    Input(shape=(100,)),
    Embedding(10000, 32),
    GRU(5, return_sequences=True),
    GRU(5),
    Dense(1, activation='sigmoid')
])

# Generate and display the visualization
plot_model(GRU_model, show_shapes=True, show_layer_names=True, to_file='gru_model.png')
display(Image(filename='gru_model.png'))
```

```python
print(GRU_model.summary())
```
